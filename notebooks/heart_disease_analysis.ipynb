### 1. Import Libraries
We import essential libraries for numerical computation (`numpy`), tabular data manipulation (`pandas`), and statistical visualization (`matplotlib` and `seaborn`).


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot styling configuration
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["font.size"] = 11
```

### 2. Load Dataset
We load `dataset/heart.csv` into a pandas DataFrame using relative paths to ensure portability across different environments.


```python
data_path = "../dataset/heart.csv"
df = pd.read_csv(data_path)
print("Dataset successfully loaded from:", data_path)
```

### 5. Dataset Shape
Determining total rows (samples) and columns (features + target).


```python
rows, cols = df.shape
print(f"Total Rows: {rows}")
print(f"Total Columns: {cols}")
```

### 6. Column Names
Examining the exact name of all 14 columns present in the dataset.


```python
for i, col in enumerate(df.columns, 1):
    print(f"{i:2d}. {col}")
```

### 7. Data Types
Checking the data types of each column to ensure all features are in numerical format for neural network consumption.


```python
df.info()
```

### 8. Statistical Summary
Summary statistics including mean, standard deviation, min, percentiles (25%, 50%, 75%), and max for all numeric attributes.


```python
df.describe().T
```

### 9. Missing Value Analysis
Checking for any missing (NaN or null) values across each column.


```python
missing_series = df.isnull().sum()
print("Missing Values per Column:")
print(missing_series)
print("\nTotal Missing Values in Dataset:", missing_series.sum())
```

### 10. Duplicate Row Analysis
Identifying repeated rows in the dataset.


```python
duplicates_count = df.duplicated().sum()
unique_count = len(df.drop_duplicates())
print(f"Total Duplicate Rows: {duplicates_count}")
print(f"Unique Rows: {unique_count}")
```

### 11. Target Distribution
Analyzing the balance between positive (1: Heart disease present) and negative (0: No heart disease) classes.


```python
fig, ax = plt.subplots(figsize=(7, 5))
counts = df["target"].value_counts()
sns.barplot(x=counts.index, y=counts.values, palette=["#3b82f6", "#ef4444"], ax=ax)
ax.set_title("Distribution of Target Variable", fontsize=14, fontweight="bold")
ax.set_xticklabels(["No Disease (0)", "Disease Present (1)"])
ax.set_ylabel("Patient Count")
for p in ax.patches:
    ax.annotate(f"{int(p.get_height())} ({p.get_height()/len(df)*100:.1f}%)",
                (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha="center", va="center", color="white", fontweight="bold", fontsize=12)
plt.tight_layout()
plt.show()
```

### 12. Age Distribution
Observing how age varies across patients and comparing the age distribution of diseased vs non-diseased individuals.


```python
plt.figure(figsize=(10, 5))
sns.histplot(data=df, x="age", hue="target", kde=True, bins=25, palette=["#3b82f6", "#ef4444"])
plt.title("Patient Age Distribution by Heart Disease Status", fontsize=14, fontweight="bold")
plt.xlabel("Age (years)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

### 13. Gender Distribution
Analyzing heart disease incidence across biological sexes (0 = Female, 1 = Male).


```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="sex", hue="target", palette=["#3b82f6", "#ef4444"])
plt.title("Heart Disease by Gender (0 = Female, 1 = Male)", fontsize=14, fontweight="bold")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.xticks([0, 1], ["Female (0)", "Male (1)"])
plt.tight_layout()
plt.show()
```

### 14. Chest Pain Type (cp) Analysis
Examining the prevalence of heart disease across the 4 chest pain categories:
- Type 0: Typical Angina
- Type 1: Atypical Angina
- Type 2: Non-anginal Pain
- Type 3: Asymptomatic


```python
plt.figure(figsize=(9, 5))
sns.countplot(data=df, x="cp", hue="target", palette=["#3b82f6", "#ef4444"])
plt.title("Heart Disease by Chest Pain Type", fontsize=14, fontweight="bold")
plt.xlabel("Chest Pain Type (cp)")
plt.ylabel("Patient Count")
plt.xticks([0, 1, 2, 3], ["Typical (0)", "Atypical (1)", "Non-anginal (2)", "Asymptomatic (3)"])
plt.tight_layout()
plt.show()
```

### 15. Serum Cholesterol (chol) Analysis
Exploring the spread and outliers of serum cholesterol across health outcomes.


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="target", y="chol", palette=["#3b82f6", "#ef4444"])
plt.title("Serum Cholesterol Level vs Heart Disease", fontsize=14, fontweight="bold")
plt.xlabel("Target Outcome")
plt.ylabel("Cholesterol (mg/dl)")
plt.xticks([0, 1], ["No Disease (0)", "Disease Present (1)"])
plt.tight_layout()
plt.show()
```

### 16. Resting Blood Pressure (trestbps) Analysis
Evaluating resting blood pressure (in mm Hg on admission to the hospital).


```python
plt.figure(figsize=(10, 5))
sns.kdeplot(data=df, x="trestbps", hue="target", fill=True, palette=["#3b82f6", "#ef4444"])
plt.title("Resting Blood Pressure Density by Disease Status", fontsize=14, fontweight="bold")
plt.xlabel("Resting Blood Pressure (mm Hg)")
plt.ylabel("Density")
plt.tight_layout()
plt.show()
```

### 17. Maximum Heart Rate (thalach) Analysis
Observing the relationship between maximum heart rate achieved during exercise stress testing and heart disease.


```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="target", y="thalach", palette=["#3b82f6", "#ef4444"])
plt.title("Maximum Heart Rate (thalach) vs Heart Disease", fontsize=14, fontweight="bold")
plt.xlabel("Target Outcome")
plt.ylabel("Maximum Heart Rate (bpm)")
plt.xticks([0, 1], ["No Disease (0)", "Disease Present (1)"])
plt.tight_layout()
plt.show()
```

### 18. Exercise-Induced Angina (exang) Analysis
Assessing angina triggered by exercise (0 = No, 1 = Yes) as a risk factor.


```python
plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="exang", hue="target", palette=["#3b82f6", "#ef4444"])
plt.title("Exercise-Induced Angina vs Disease Status", fontsize=14, fontweight="bold")
plt.xlabel("Exercise-Induced Angina (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.xticks([0, 1], ["No (0)", "Yes (1)"])
plt.tight_layout()
plt.show()
```

### 19. Feature Correlation Heatmap
Computing Pearson correlation matrix across all 14 features to identify collinearity and relationships with the target.


```python
plt.figure(figsize=(12, 10))
corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, square=True, linewidths=0.5)
plt.title("Pearson Correlation Heatmap", fontsize=15, fontweight="bold", pad=12)
plt.tight_layout()
plt.show()
```

### 20. Multivariate Feature-vs-Target Visualizations
Pairing key physiological markers (Maximum Heart Rate vs Age, ST Depression vs Slope) stratified by disease outcome.


```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Max Heart Rate vs Age
sns.scatterplot(data=df, x="age", y="thalach", hue="target", palette=["#3b82f6", "#ef4444"], alpha=0.7, ax=axes[0])
axes[0].set_title("Max Heart Rate vs Age by Heart Disease Status", fontsize=13, fontweight="bold")
axes[0].set_xlabel("Age (years)")
axes[0].set_ylabel("Max Heart Rate (thalach)")

# ST Depression (oldpeak) vs Slope
sns.boxplot(data=df, x="slope", y="oldpeak", hue="target", palette=["#3b82f6", "#ef4444"], ax=axes[1])
axes[1].set_title("ST Depression (oldpeak) by ST Slope", fontsize=13, fontweight="bold")
axes[1].set_xlabel("ST Slope (0: Upsloping, 1: Flat, 2: Downsloping)")
axes[1].set_ylabel("ST Depression (oldpeak)")

plt.tight_layout()
plt.show()
```
